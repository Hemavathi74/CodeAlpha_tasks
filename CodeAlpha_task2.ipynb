```python
categorical_columns = [
    'Airline',
    'Travel_Class',
    'Departure_City',
    'Payment_Method'
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts())
```

```python
df['Airline'].value_counts().plot(kind='bar')
plt.title('Bookings by Airline')
plt.xlabel('Airline')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
df['Departure_City'].value_counts().plot(kind='bar')
plt.title('Bookings by Departure City')
plt.xlabel('Departure City')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
df['Travel_Class'].value_counts().plot(kind='bar')
plt.title('Bookings by Travel Class')
plt.xlabel('Travel Class')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
df.groupby('Airline')['Ticket_Price'].mean().plot(kind='bar')
plt.title('Average Ticket Price by Airline')
plt.xlabel('Airline')
plt.ylabel('Average Ticket Price')
plt.show()
```

# EDA Findings

1. **Airline:** B Airlines has the highest number of bookings.
2. **Departure City:** Mumbai has the highest number of bookings.
3. **Travel Class:** Premium Economy has the highest number of bookings.
4. **Average Ticket Price:** Air One has the highest average ticket price.
5. **Flight Duration vs Ticket Price:** The scatter plot shows the relationship between flight duration and ticket price.
